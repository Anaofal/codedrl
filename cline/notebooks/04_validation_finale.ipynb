```python
# === IMPORTS SUPPLÉMENTAIRES POUR L'ANALYSE ===
import scipy.stats
import seaborn as sns
from scipy.stats import norm, t
from scipy.optimize import minimize
import warnings
warnings.filterwarnings('ignore')  # Pour une sortie plus propre

print("Imports supplémentaires chargés ✅")
```

```python
# === FONCTIONS D'ANALYSE STATISTIQUE ===
def calculate_distribution_stats(returns):
    """Calcule les statistiques de distribution complètes"""
    if len(returns) == 0:
        return None
    
    stats = {
        'mean': np.mean(returns),
        'median': np.median(returns),
        'std': np.std(returns),
        'skewness': scipy.stats.skew(returns),
        'kurtosis': scipy.stats.kurtosis(returns),
        'min': np.min(returns),
        'max': np.max(returns),
        'q1': np.percentile(returns, 25),
        'q3': np.percentile(returns, 75),
        'iqr': np.percentile(returns, 75) - np.percentile(returns, 25)
    }
    
    # Calcul des mesures de risque
    stats.update({
        'var_95': np.percentile(returns, 5),
        'var_99': np.percentile(returns, 1),
        'cvar_95': np.mean(returns[returns <= stats['var_95']]),
        'cvar_99': np.mean(returns[returns <= stats['var_99']])
    })
    
    # Test de normalité
    _, stats['normality_pvalue'] = scipy.stats.normaltest(returns)
    
    return stats

def detect_outliers_multiple(returns, methods=['zscore', 'iqr', 'mad']):
    """Détecte les outliers avec plusieurs méthodes"""
    outliers = np.zeros(len(returns), dtype=bool)
    
    if 'zscore' in methods:
        z_scores = np.abs((returns - np.mean(returns)) / np.std(returns))
        outliers |= z_scores > 3
        
    if 'iqr' in methods:
        q1, q3 = np.percentile(returns, [25, 75])
        iqr = q3 - q1
        outliers |= (returns < (q1 - 1.5 * iqr)) | (returns > (q3 + 1.5 * iqr))
        
    if 'mad' in methods:
        median = np.median(returns)
        mad = np.median(np.abs(returns - median))
        outliers |= np.abs(returns - median) > (3 * mad)
    
    return outliers

def fit_distribution(returns):
    """Ajuste plusieurs distributions et trouve la meilleure"""
    distributions = [
        ('normal', norm),
        ('student', t)
    ]
    
    best_fit = {
        'distribution': None,
        'params': None,
        'aic': np.inf
    }
    
    for name, dist in distributions:
        try:
            params = dist.fit(returns)
            log_likelihood = np.sum(dist.logpdf(returns, *params))
            k = len(params)
            n = len(returns)
            aic = 2 * k - 2 * log_likelihood
            
            if aic < best_fit['aic']:
                best_fit.update({
                    'distribution': name,
                    'params': params,
                    'aic': aic
                })
        except:
            continue
    
    return best_fit

def plot_return_analysis(returns, asset_name, cleaned_returns=None):
    """Génère une analyse visuelle complète des rendements"""
    fig = plt.figure(figsize=(15, 10))
    gs = plt.GridSpec(2, 3, figure=fig)
    
    # Histogramme et KDE
    ax1 = fig.add_subplot(gs[0, 0])
    sns.histplot(returns, kde=True, ax=ax1)
    if cleaned_returns is not None:
        sns.kdeplot(cleaned_returns, color='red', ax=ax1)
    ax1.set_title(f"Distribution des rendements - {asset_name}")
    
    # QQ-Plot
    ax2 = fig.add_subplot(gs[0, 1])
    scipy.stats.probplot(returns, dist="norm", plot=ax2)
    ax2.set_title("QQ-Plot vs. Normal")
    
    # Box-plot
    ax3 = fig.add_subplot(gs[0, 2])
    sns.boxplot(y=returns, ax=ax3)
    ax3.set_title("Box-plot")
    
    # ACF des rendements
    ax4 = fig.add_subplot(gs[1, 0])
    pd.plotting.autocorrelation_plot(pd.Series(returns), ax=ax4)
    ax4.set_title("Autocorrélation")
    
    # ACF des rendements absolus (volatilité)
    ax5 = fig.add_subplot(gs[1, 1])
    pd.plotting.autocorrelation_plot(pd.Series(np.abs(returns)), ax=ax5)
    ax5.set_title("Autocorrélation (|rendements|)")
    
    # Volatilité mobile
    ax6 = fig.add_subplot(gs[1, 2])
    rolling_std = pd.Series(returns).rolling(window=20).std()
    rolling_std.plot(ax=ax6)
    ax6.set_title("Volatilité mobile (20 périodes)")
    
    plt.tight_layout()
    return fig

print("Fonctions d'analyse statistique chargées ✅")
```

```python
# === FONCTION PRINCIPALE DE VALIDATION DES RENDEMENTS ===
def validate_and_analyze_returns(sim_returns, assets, 
                               max_return=5.0, min_return=-1.0,
                               outlier_methods=['zscore', 'iqr', 'mad']):
    """
    Fonction complète de validation et d'analyse des rendements simulés
    
    Parameters:
    -----------
    sim_returns : np.array
        Matrice des rendements simulés (n_simulations x n_assets)
    assets : list
        Liste des noms des actifs
    max_return : float
        Rendement maximal accepté
    min_return : float
        Rendement minimal accepté
    outlier_methods : list
        Méthodes de détection des outliers à utiliser
    
    Returns:
    --------
    dict : Résultats complets de l'analyse
    """
    results = {
        'cleaned_returns': sim_returns.copy(),
        'asset_stats': {},
        'global_stats': {},
        'outliers': {},
        'distribution_fits': {}
    }
    
    print("\n=== ANALYSE DÉTAILLÉE DES RENDEMENTS SIMULÉS ===")
    
    # Analyse par actif
    for i, asset in enumerate(assets):
        print(f"\n📊 Analyse de {asset}")
        returns = sim_returns[:, i]
        
        # 1. Détection des valeurs invalides
        invalid_mask = np.isnan(returns) | np.isinf(returns)
        if np.any(invalid_mask):
            print(f"⚠️ {np.sum(invalid_mask)} valeurs invalides détectées")
            valid_returns = returns[~invalid_mask]
        else:
            valid_returns = returns
            
        # 2. Statistiques initiales
        print("\n▶️ Statistiques avant nettoyage:")
        initial_stats = calculate_distribution_stats(valid_returns)
        if initial_stats:
            for key, value in initial_stats.items():
                if isinstance(value, float):
                    if key in ['mean', 'median', 'std', 'min', 'max', 'var_95', 'cvar_95']:
                        print(f"   - {key}: {value:+.2%}")
                    else:
                        print(f"   - {key}: {value:.4f}")
                        
        # 3. Détection des outliers
        outliers = detect_outliers_multiple(valid_returns, methods=outlier_methods)
        results['outliers'][asset] = outliers
        
        # 4. Nettoyage
        if np.any(outliers):
            valid_indices = ~outliers
            if np.any(valid_indices):
                median_return = np.median(valid_returns[valid_indices])
            else:
                median_return = 0
            results['cleaned_returns'][outliers, i] = median_return
            
        # 5. Statistiques après nettoyage
        print("\n▶️ Statistiques après nettoyage:")
        cleaned_stats = calculate_distribution_stats(results['cleaned_returns'][:, i])
        results['asset_stats'][asset] = {
            'before': initial_stats,
            'after': cleaned_stats
        }
        
        if cleaned_stats:
            for key, value in cleaned_stats.items():
                if isinstance(value, float):
                    if key in ['mean', 'median', 'std', 'min', 'max', 'var_95', 'cvar_95']:
                        print(f"   - {key}: {value:+.2%}")
                    else:
                        print(f"   - {key}: {value:.4f}")
        
        # 6. Ajustement de distribution
        fit_result = fit_distribution(results['cleaned_returns'][:, i])
        results['distribution_fits'][asset] = fit_result
        print(f"\n▶️ Meilleure distribution: {fit_result['distribution']}")
        
        # 7. Visualisation
        fig = plot_return_analysis(returns, asset, results['cleaned_returns'][:, i])
        plt.show()
        
    # Statistiques globales
    print("\n=== STATISTIQUES GLOBALES ===")
    results['global_stats'].update({
        'total_simulations': len(sim_returns),
        'valid_simulations': np.sum(np.all(~np.isnan(results['cleaned_returns']) & 
                                         ~np.isinf(results['cleaned_returns']), axis=1)),
        'correlation_matrix': np.corrcoef(results['cleaned_returns'].T)
    })
    
    print(f"Nombre total de simulations: {results['global_stats']['total_simulations']}")
    print(f"Simulations valides: {results['global_stats']['valid_simulations']}")
    
    # Affichage de la matrice de corrélation
    plt.figure(figsize=(10, 8))
    sns.heatmap(pd.DataFrame(results['global_stats']['correlation_matrix'], 
                            index=assets, columns=assets),
                annot=True, cmap='coolwarm', center=0)
    plt.title("Matrice de Corrélation des Rendements Nettoyés")
    plt.tight_layout()
    plt.show()
    
    return results

print("Fonction principale de validation chargée ✅")
```

```python
# === VALIDATION ET NETTOYAGE DES RENDEMENTS SIMULÉS ===
print("\n" + "="*60)
print("🧹 VALIDATION ET NETTOYAGE DES RENDEMENTS")
print("="*60)

def validate_returns(sim_returns, assets, max_return=5.0, min_return=-1.0):
    """
    Valide et nettoie les rendements simulés
    max_return: rendement maximal accepté (500%)
    min_return: rendement minimal accepté (-100%)
    """
    print("\n--- Analyse des rendements par actif ---")
    
    valid_mask = np.ones(sim_returns.shape[0], dtype=bool)
    cleaned_returns = sim_returns.copy()
    
    for i, asset in enumerate(assets):
        returns = sim_returns[:, i]
        
        # Vérification des valeurs invalides
        invalid_mask = np.isnan(returns) | np.isinf(returns)
        extreme_mask = (returns > max_return) | (returns < min_return)
        outliers = invalid_mask | extreme_mask
        
        # Statistiques avant nettoyage (avec gestion des valeurs invalides)
        print(f"\n{asset}:")
        print(f"  Avant nettoyage:")
        if np.any(invalid_mask):
            print(f"   ⚠️ Valeurs invalides détectées: {np.sum(invalid_mask)} NaN/Inf")
            valid_returns = returns[~invalid_mask]
        else:
            valid_returns = returns
            
        if len(valid_returns) > 0:
            print(f"   - Moyenne: {np.mean(valid_returns):+.2%}")
            print(f"   - Std: {np.std(valid_returns):+.2%}")
            print(f"   - Min: {np.min(valid_returns):+.2%}")
            print(f"   - Max: {np.max(valid_returns):+.2%}")
        else:
            print("   ❌ Aucune valeur valide!")
        
        # Nettoyage des valeurs aberrantes
        if np.any(outliers):
            valid_mask &= ~outliers
            # Remplacer les valeurs aberrantes par la médiane des valeurs valides
            median_return = np.median(returns[~outliers]) if np.any(~outliers) else 0
            cleaned_returns[outliers, i] = median_return
            
        # Statistiques après nettoyage
        clean_returns = cleaned_returns[:, i]
        print(f"  Après nettoyage:")
        print(f"   - Moyenne: {np.mean(clean_returns):+.2%}")
        print(f"   - Std: {np.std(clean_returns):+.2%}")
        print(f"   - Min: {np.min(clean_returns):+.2%}")
        print(f"   - Max: {np.max(clean_returns):+.2%}")
        print(f"   - Valeurs corrigées: {np.sum(outliers)}/{len(returns)} ({np.mean(outliers):+.2%})")
        
        if np.sum(outliers) > 0:
            print(f"   ⚠️ Actif potentiellement problématique!")
    
    return cleaned_returns

# Test avec les rendements précédents
print("\n--- Test de validation des rendements ---")
try:
    sim_returns, _, valid_assets = copula_simulator.simulate_period_returns(
        test_assets, test_date, period_days=7
    )
    
    if sim_returns is not None:
        # Validation et nettoyage
        clean_returns = validate_returns(sim_returns, valid_assets)
        
        # Visualisation avant/après avec échelle adaptative
        fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(15, 10))
        
        # Avant nettoyage - échelle normale
        ax1.boxplot(sim_returns, labels=valid_assets)
        ax1.set_title("Distribution Avant Nettoyage")
        ax1.set_ylabel("Rendement")
        ax1.tick_params(axis='x', rotation=45)
        ax1.grid(True)
        
        # Avant nettoyage - échelle log
        ax2.boxplot(np.clip(np.abs(sim_returns), 1e-10, None), labels=valid_assets)
        ax2.set_yscale('log')
        ax2.set_title("Distribution Avant Nettoyage (échelle log)")
        ax2.tick_params(axis='x', rotation=45)
        ax2.grid(True)
        
        # Après nettoyage - échelle normale
        ax3.boxplot(clean_returns, labels=valid_assets)
        ax3.set_title("Distribution Après Nettoyage")
        ax3.set_ylabel("Rendement")
        ax3.tick_params(axis='x', rotation=45)
        ax3.grid(True)
        
        # Après nettoyage - échelle log
        ax4.boxplot(np.clip(np.abs(clean_returns), 1e-10, None), labels=valid_assets)
        ax4.set_yscale('log')
        ax4.set_title("Distribution Après Nettoyage (échelle log)")
        ax4.tick_params(axis='x', rotation=45)
        ax4.grid(True)
        
        plt.tight_layout()
        plt.show()
        
        print("\n--- Statistiques globales ---")
        print(f"Nombre total de simulations: {len(sim_returns)}")
        print(f"Nombre de simulations valides: {np.sum(np.all(~np.isnan(clean_returns) & ~np.isinf(clean_returns), axis=1))}")
        
except Exception as e:
    print(f"❌ Erreur lors de la validation: {e}")

# Suggestion d'amélioration de la copule
print("""
🔧 Suggestions d'amélioration pour la copule:
1. Ajouter des bornes de validation dans la simulation GJRGARCH
2. Utiliser une fenêtre glissante adaptative par actif
3. Implémenter une détection d'instabilité plus stricte
4. Ajouter un pré-traitement des données aberrantes
5. Mettre en place un système de fallback pour les actifs instables
""")
```
